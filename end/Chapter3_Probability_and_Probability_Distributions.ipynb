Probability deals with uncertainty. Mathematically, it allows us to numerically describe or quantify this uncertainty.

$\bf{Different \, interpretations \, of \, Probability}$

- (1) $\bf{Classical \, Interpretation \, of \, Probability}$

This is associated with the games of chances. For example, while flipping an unbiased coin what is the chance to get Heads? or what is the probability of drawing an ace from a standard deck of cards?

The numerical values of the probabilities associted to the above questions arise from the nature of the game. For example: the flipping of a coin has two possible outcomes (head or tail), thus, probabiity of getting a head = 1/2.
Similarly, there are 4 ace cards in a standard deck of 52 cards, so the probability in drawing an ace card is 4/52.

# $P(E) = \frac{Number\, of \,favourable \,cases}{Total \, number \, of \, possible \, outcomes}$

This definition was given by James Bernoulli who was the first person to obtain a quantitative measure of uncertainty.

- (2) $\bf{Relative \,  frequency \, concept \, of \, probability}$

It is an $\bf{empirical \,approach}$ to describe probability of an event. For example, if an experiment is repeated a large number of times and say an event occurs 30% of the time, then the probability for the event is given by

$P(E) \sim n_e/n$

where $n$ is the number of experiments and $n_e$ is the number of experiments that gave desired E event as its outcome. Note, these experiments can also be simulations.

- (3) $\bf{Personal \, or \, subjective \, probability}$

It comes in a situation where there is no possibility of repeating an experiment. In such a case, it is based on the knowledge or the subjective belief of the various factors involved in the experiment allows to give an idea how likely an event can occur. The problem with subjective probabilities is that they can vary from person to person and they cannot be checked.




# Outcome spaces and Events

# Brief discussion on Set Theory

Probability theory works with sets in a given space. A set is a collection of elements and a space is the collection of all elements under consideration. For example, $A_1$={1}, $A_2$={1,4,20,52}, and $A_3$={10,25,70}, are all sets contained in the space of natural numbers, $\mathbb{N}=\{0,1,…\}$.

In other words, this is denoted as $A_1 \subset \mathbb{N}$, $A_2 \subset \mathbb{N}$ etc

Among the above sets, $A_1$ is referred to as a point set or atomic set as the set contains only 1 element. The entire space (for eg: $\mathbb{N}$) itself is always a valid set. Also, the empty set or null set, $\varnothing$, which contains no elements at all, is always a valid set.

Sets are often defined by an inclusion criterion. For example

# $A=\{x \in \Omega∣f(x)=0\}$,

which is read as “the set of elements $x$ in the space $\Omega$ such that the condition $f(x)=0$ holds”.

Consider for example sets:

$A_1 = \{2, 5, 3, 4\}$ and $A_2 = \{7, 8, 9, 10\}$  which can be written as

$A_1 = \{x \in \mathbb{N}|2 \le x \le 5 \}$ and

$A_2 = \{x \in \mathbb{N}|6 \le x \le 10\}$




$\bf{Set \, Operations}$

Three natural operations between sets: set complement, set union, and set intersection.

- $\bf{Set \, Complement}$

For a given set $A$, complement of set $A$ given as $A^{c}$ includes all of the elements not in the set $A$,

# $A^{c} = \{x \in \Omega∣x \notin A\}$

For example the complement of $A_1^{c} = \{ x \in \mathbb{N}| x < 2 \, OR \, x > 5\} $


 - $\bf{Set \, Union}$
 It is a binary operator mapping into a third set.

 # $A_1 \cup A_2 = \{x \in \Omega∣x \in A_1 \, OR\,  x \in A_2 \}$


- $\bf{Set \, Intersection}$
It is also a binary operator that maps into a third set such that the third set includes elements that are present in both the sets

# $A_1 \cap A_2 = \{x \in \Omega∣x \in A_1 \, AND\,  x \in A_2\}$



$\bf{Experiment}$

An experiment is any action that can have a set of possible results where the actually occurring result cannot be predicted with certainty prior to the action.

$\bf{Outcome \, Space \, or \, Sample \, Space \, (\Omega)}$

The outcome space consists of each distinct possible result is referred to as outcome.

$\bf{Event, \, (E)}$

It is a subset of outcome/sample space.
The collection of preferred outcomes is referred to as an event.

Thus, the probability of an event, $E$ is estimated as

# $P(E) = \frac{N_e}{N}$

where $N_e$ is the number of outcomes favourable to event, $E$ and $N$ is the total number of possible outcomes.

In simplest cases, such as toss of a coin where $\Omega = \{H, T \}$ the P(H) = 1/2 and P(T) = 1/2. Another example, dice whose $\Omega = \{1, 2, 3, 4, 5, 6 \}$ and the probability of each of the outcome is equal to 1/6.

Now, let us consider we have three different unbiased coins the outcome space is

$\Omega = \{HHH,\, HHT,\, HTH,\, HTT,\, THH,\, THT,\, TTH,\, TTT \}$

Here, if we ask the question what is the probability to get P(HHH) = 1/8, as each of the outcome is equally probable.

But there can also be scenarios where each of the outcome in the space is not equally probable. Example, if for the above case, we are interested to know after the three coins are tossed, how many heads appear. In that case, the outcome space is

$\Omega = \{0, 1, 2, 3\}$

Now the probability to get an outcome of 0 heads is not equal to the probability to get an outcome of 2 heads right?
Let us see,

E = 0  is possible when TTT occurs

E = 1 is possible when HTT or THT or TTH occurs

E = 2 is possible when HHT or HTH or THH occurs

E = 3  is possible when HHH occurs

P(0) = P(3) = 1/8

P(1) = P(2) = 3/8

In such cases, a probability model assigns a weight (p_i) to each possible outcome and thus,

# $ P(E) = \sum_{i:e_i \in E} p_i $

In other words, in probability theory, weight given to an event is how probable that event is to happen.   


The sample space $\Omega$ can be $\bf{discrete \, or \,  it \, can \,  be \,  continous}$. For example in astronomy, countable spaces are such as the number of photons detected, number stars in a galaxy etc, whereas uncountable spaces are such as variability in a quasar emission or a GRB light curve, background noise in an image constructed from interferometry etc.

# Axioms of Probability

A probability space consists of $(\Omega, \, E, \, P)$ where $\Omega$ is outcome/ sample space, $E$ is a class of events of interest and $P$ is the function that assigns a probability to each event in the class $E$. They obey three axioms of probability:

- $\bf{Axiom \, 1:}$  $0 \le P(E) \le 1$ for all events in $E$.

- $\bf{Axiom \, 2:}$ $P(\Omega) = 1$

- $\bf{Axiom \, 3:}$ For mutually exclusive events say $A_1$, $A_2$, $A_3$, ... etc,

$P(A_1 \cup A_2 \cup A_3 ....)$ = $P(A_1) \, + \, P(A_2) \, + \, P(A_3)\, +\, ....$

# Conditional Probability

Consider an experiment gives out $m$ equally likely outcomes and A and B are two events. If we are given the information that B has happened, what is the probability that A has happened in light of the new knowledge? In other words, what is the probability of A provided B has already occured?

Let us take the example of a dice. We want to look at the probability of getting numbers $\{1, 2, 3\}$
when we throw the dice, lets refer to this as event A? Now, lets say in the first throw, we got an even number which we can refer to as event B. Now given this new information of B, what is the probability of event A?

# $P(A|B) = \frac{P(A \, \cap  \,  B)}{P(B)}$

**The right hand side of the equation tells us what fraction of the probability of the occurance of B intersects with the occurance of A.**

The events of our interest are $A = \{1,2,3 \}$ and $B = \{2,4,6 \}$ and the outcome/ sample space
$\Omega = \{1,2,3,4,5,6\}$.

Now, P(B) = 3/6 = 1/2.

what is the P(A $\cap$ B)? It is =  1/6.

P(A|B) = 1/6 divided by 1/2 = 1/3.

P(A) = 3/6 = 1/2 =  0.5

Given the information, we got the event B, what is the probability to get A, this comes to be P(A|B) = 1/3 = 0.33.

In the above example, P(A|B) is referred to as $\bf{conditional \, probability}$.

P(B) is referred to as $\bf{unconditional \, or \, marginal \, probability}$.

$P(A \, \cap  \,  B)$ is referred to as the $\bf{joint \, probability}$.

# Independent Events

The occurrence of event A is not dependent on the occurrence of event B or

# $P(A|B) = P(A)$

Such events A and B are referred to as independent events.

For the case of three events, A, B, C are independent if all the following four conditions are satisfied:

# $ P(A \cap B \cap C) = P(A) \, P(B)\, P(C)$ ,
# $P(A \cap B) = P(A)\, P(B)$ ,
# $P(B \cap C) = P(B)\, P(C)$,
# $P(A \cap C) = P(A)\, P(C)$.

# Dependent Events

The occurrence of A depends on the occurrence of B or

# $P(A|B) \neq P(A)$

Such events A and B are referred to as dependent events.

# Mutually Exclusive Events
Events are said to be mutually exclusive if no two or more of them can happen simultaneouly in the same experiment.
# $P(A \cap B) = 0$

# Bayes theorem

The theorem was formulated by the British Mathematician, Thomas Bayes in the 18th century.
As we had seen earlier, the conditional probability gives a revised probability giving a new information or condition. This is the essence of Bayes theorem. In most cases, we begin with a prior probability of an event $A$ and later when we get some information of the event, we update the prior probability to a get a revised probabilities referred to as posterior probabilities.


Bayes theorem: If $A_1, A_2, A_3,....., A_n$ are mutually disjoint (mutually exclusive) events with $P(A_i) \neq 0$, where $i = 1, 2, 3, ....n$, then for any arbitrary event A which is a subset of $\Omega = \cup_{i=1}^{n} A_i $ such that P(A) > 0, we have

# $ P(A_i|A) = \frac{P(A_i) \, P(A|A_i)}{P(A) = \sum_{i=1}^n \, P(A_i) \, P(A|A_i)}$

## There is some number $i$ possible, mutually exclusive, underlying events $A_1, . . . , A_i$, they are referred to as the states of nature.
## The unconditional probability, $P(A_i)$ is referred to as the prior probability.
## The $A$ mutually exclusive event is referred to as the observable event (or the information or condition).
## The conditional probability of the observable event given the state of nature, $P(A|A_i)$, are also specified, and these probabilities are called likelihoods.

## The problem is to find the posterior probability $P(A_i|A)$.

Prior and posterior refer to probabilities before and after observing an event A.


Putting these in words, The probability that a belief ($B$) is true given a new evidence ($E$) which is $P(B|E)$ equals the probability that the belief is true regardless of that evidence i.e  $P(B)$,  times the probability that the evidence is true given that the belief is true i.e $P(E|B)$, divided by the probability that the evidence is true regardless of whether the belief is true i.e $P(E)$.

# $ P(B|E) = \, \frac{P(E|B) P(B)}{P(E)}$

# Random Variable

In many situations, it is not necessary that we focus on the entire outcome space, it may be sufficient that we concentrate on a summary of outcomes relevant to the problem at hand. For example, in astronomy, instead of looking at every single photon that is detected by the instrument, it is sufficient that we look at the average of the photons in a spectral band integrated over some time.

These real-valued functions on the outcome space or sample space are called random variables, $X$. In other words, it is a mathematical formulation or function that provides the connection between the outcome space, $\Omega$ to the measured space, $E$.

In the above example of set A
### $A=\{x \in \Omega∣f(x)=0\}$,
here $f(x)$ is the random variable and when we put a constraint or a condition on this function i.e $f(x)=0$ allows us to generate the outcomes in the set $A$.

A random variable is called a $ \bf{discrete \,  random \, variable}$ if it maps a sample space to a countable set (e.g. the integers) with each value in the range having probability greater than or equal to zero.

A random variable is called a $ \bf{continous \,  random \, variable}$ if it maps a sample space that is uncountably infinte (thereby usually presented as an interval).

### In practice, the random variable $X$ is not directly accessible; only its realizations are observed (data). From these data, we infer the probability distribution that governs $X$.


# Probability Distribution

Presenting the probabilities of the output ranges of the real-valued random variable $X$ gives the probability distribution of $X$.

The distribution corresponding to the discrete random variable, $X$ is called as $\bf{discrete \, probability \,  distribution \, or \, probability \, mass \, function}$.

For continous random variables, are when we generally use the $\bf{probability \, density \, distribution}$ where the probability are assigned per unit length of the interval.


# Moments of a distribution
The moments of a distribution allow us to understand how the probability of the random variable is distributed. In other words, they tell us the shape of the probability distribution.

Why are they called moments?

It has some relation to physics for example the moment of inertia of an object is related to how the mass of the object is distributed about its axis of rotation.  

- Moments quantify three parameters of distributions: $\bf{location, \, shape, \, and \, scale}$.

A distribution’s $\bf{ location}$ refers to where the center of mass of the distribution is along the x-axis. By convention, a mean-centered distribution has a center of mass at zero.

$\bf{The\, scale}$ refers to how much is the deviation or spread in the distribution. Scale stretches or compresses a distribution along the x-axis.

$\bf{The\,  shape}$ of a distribution refers to its overall geometry: that is if the distribution is bimodal, asymmetric, heavy-tailed etc.

The kth moment of a function $f(x)$ about a non-random value $c$ is

# $ E\, [(X−c)^k] = \int_{-\infty}^{+\infty}(x−c)^k \, f(x)\, dx $.

# When $c=0$ it is referred to as the raw moment. If $c = \, mean \, = \mu$, then it is referred to as central moments. If it is $\left(\frac{X-\mu}{\sigma}\right)^k$, it is referred to as the standardised moment.   

Central moments are defined when $c = \mu$, the mean of the distribution is then defined as

# $ E\, [(X−\mu)^k] = \int_{-\infty}^{+\infty}(x−\mu)^k \, f(x)\, dx $.

# - The $\bf{zeroth\,  central\,  moment}$  (i.e $k=0$) is always equal to 1

# $ E\, [(X−\mu)^0] = \int_{-\infty}^{+\infty}(x−\mu)^0 \, f(x)\, dx $ = 1.

This tells that the probability distributions are normalised quantities.

# - The $\bf{first\,  central \, moment}$ (i.e $k=1$) is equal to zero.

# $ E\, [(X−\mu)^1] = \int_{-\infty}^{+\infty}(x−\mu)^1 \, f(x)\, dx $ = 0.

# From the $\bf{first \, raw\, moment}$, we can find the location of the centre of mass of the distribution that is the $\bf{mean}$.


# - The $\bf{Second\, central \, moment}$ (i.e $k=2$) tells us about the spread of the distribution that is $\bf{variance}$.

# $ E\, [(X−\mu)^2] = \int_{-\infty}^{+\infty}(x−\mu)^2 \, f(x)\, dx $.

```python
import numpy.random as npr
import matplotlib.pyplot as plt

# `X` has larger values on average, but `Y` has higher variability in values.
X = npr.normal(0, 1, size=10000)
Y = npr.normal(0, 2, size=10000)


plt.figure()
plt.hist(Y,bins=100,label='Y')
plt.hist(X,bins=100,label='X')
plt.legend(loc='best')
plt.show()


# But `Y` has a larger second central moment (variance).
mx = X.var()
my = Y.var()
print('Variance of X =',f'{mx:.2f}','Variance of Y = ',f'{my:.2f}')
```

# - The $\bf{Third \, standardised \, central \, moment}$ gives us the $\bf{skewness}$ in the data which measures the relative size of the two tails of a distribution.

# $ E\, [\left(\frac{X-\mu}{\sigma}\right)^3] = \int_{-\infty}^{+\infty}\left(\frac{X-\mu}{\sigma}\right)^3 \, f(x)\, dx $.

where $\left(\frac{X-\mu}{\sigma}\right) = Z$ which is referred to as standard score or z-score.

In other words, data points less than a standard deviation from the mean contribute very little to the final calculation of skewness. Since the cubic function preserves sign, if both tails are balanced, the skewness is zero. Otherwise, the skewness is positive for longer right tails and negative for longer left tails.

Note: It cannot be said that skewness tells us about the symmetry of assymmetry in the distribution. Instead it is best to say that skewness measures the relative size of the two tails of the distribution.

# Random Sample
A sample of $n$ measurements selected from a population is said to be a random sample if every different sample of size $n$ from the population has an equal probability of being selected.
In other words, a random sample is a collection of independent observations, each drawn from the same probability distribution.

# Discrete Probability Distributions

# 1. Bernoulli Distribution
Suppose an experiment results in a success or failure (True or False) and we define $X =1$ if the outcome is a success and X =0 if the outcome is a failure.

X is called a binary variable or dichotomous variable. That is there are only two possible outcomes for the experiment. Each of the outcome has fixed probability of occurance and each trial of the experiment is independent of each other.

The distribution is such that

# $P(X = 1) \, = p$,  and

# $P(X=0) \, = 1- p$

is called the Bernoulli distribution, where 0 < p < 1.

**For example, coin toss, online purchase: customer purchases or not, manufactoring defect: the product is defective or not, will it rain today or not etc.**

```python
from scipy.stats import bernoulli
import seaborn as sb
import numpy as np

p=0.6
data_bern = bernoulli.rvs(size=100,p=p)
#print(data_bern)
ax = sb.displot(data_bern,
                  kde=False,
                  color='crimson')
                  #hist_kws={"linewidth": 50,'alpha':1})
ax.set(xlabel='Bernouli', ylabel='Frequency')

mean, var, skew, kurt = bernoulli.stats(p=p, moments='mvsk')
print(np.mean(data_bern),np.var(data_bern))
print(mean, var, skew, kurt)

print(moments(data_bern))
```

# 2. Binomial distribution

Suppose the above Bernoulli experiment is repeated $n$ times independently and the number of successes are denoted by X. Then the distribution of X is given by

# $ P(X =i) =  \frac{n!}{i! (n-i)!} p^i (1 - p)^{n-i}$  

where $i = 0, 1, .....,n$. In this case, the random variable X is said to have the binomial distribution and is denoted by $X \sim  Bin(n, p)$.

# Mean of $\mu = n p$ and Standard Deviation, $\sigma = \sqrt{np (1-p)}$.

```python
from scipy.stats import binom
import matplotlib.pyplot as plt
# setting the values of n and p
n = 10
p= 0.4
#Generate the data
data_binom = binom.rvs(size=5000,n=10, p=0.4)
data_binom2 = binom.rvs(size=100,n=10, p=0.4)


plt.figure()

counts,edges,patches=plt.hist(data_binom,bins=10,density=False,facecolor='blue',alpha=0.5)
#plt.hist(data_binom,bins=5,density=False,facecolor='red',alpha=0.5)
plt.tick_params(labelsize=15)
plt.tick_params(axis='x', labelsize=18)
plt.tick_params(axis='y', labelsize=18)
plt.xlabel('X',fontsize=15)
plt.show()

mean, var, skew, kurt = binom.stats(n, p, moments='mvsk')
print(mean, var, skew, kurt)
print(moments(data_binom))
```

# 3. Poisson Distribution

Poisson distribution occurs when there are events which do not occur as outcomes of a definite number of trials (unlike in binomial distribution) of an experiment but which occur at random points of time and space wherein our interest lies only in the number of occurences of the event, not in its non-occurences.  

A random variable $X$ is said to have a Poisson distribution, denoted $X \sim Poi(\lambda)$,with rate $\lambda>0$ if

# $
P(X=i) =
\begin{equation*}
  \frac{\lambda^i}{i!} e^{-\lambda}, for \, i =0, 1,...
\end{equation*}
$

Here, $i$ is the number of occurences.

# $\mu = \operatorname {E} (X) = \lambda $

# Limit Theorems
Probability theory has many powerful mathematical results which establish or constrain properties of random variables.
The uncertainty associated at the micro-level say for example the uncertainty associated with a single experiment or event, leads to deterministic behaviour at the macro level. It is a consequence of Law of large numbers.

# (a) Law of Large Numbers
Let $X_1$,$X_2$,... be a sequence of independent random variables with a common distribution and
$E[|X_1|] < \infty$.

# $
\bar{X}_n =
\begin{equation*}
\frac{1}{n} \sum_{i=1}^{n} X_i
\end{equation*}
$

# For $\epsilon > 0$, in the limit of $n$, $ \lim_{n\to\infty}P(|\bar{X}_n - \mu|> \epsilon) \rightarrow 0$

**The probability of the sample mean deviating significantly from the true mean reduces as n increases.**

# That is $\bar{X}_n$ (sample mean) tends to $\mu$ (population mean).

It is important to note that this result is valid for discrete random variables, continuous random variables, and for general random variables.

# Sampling Distribution

# The numerical descriptive measures of a population (parameters) is $\mu$ and $\sigma$.
# The numerical descriptive measures of a sample (statistics) is $\bar{y}$ and $s$.

A random sample mean, $\bar{y}$ need not be equal to the population mean, $\mu$ unless the sample size is very large.
The sample statistic itself is a random variable, it is subject to random variation because it is based on a random sample of measurements selected from the population of interest. Thus, the sample statistic will also have a probability distribution. We call the probability distribution of a sample statistic the sampling distribution of that statistic. It includes all possible values of the sample statistic.  

# (b) Central Limit Theorem for $\bar{y}$
Let $\bar{y}$ denote the sample mean computed from a random sample and $n$ such measurements from a population having a mean, $\mu$, and finite standard deviation, $\sigma$. Let $\mu_{\bar{y}}$ and $\sigma_{\bar{y}}$ denote the mean and standard deviation of the sampling distribution of $\bar{y}$, respectively. Based on n repeated random samples from the population, we can conclude the following:
# 1. $\mu_{\bar{y}} = \mu$
# 2. $\sigma_{\bar{y}} = \frac{\sigma}{\sqrt{n}}$
## 3. When $n$ is large, the sampling distribution of $\bar{y}$ will be approximately normal (with the approximation becoming more precise as $n$ increases).
## 4. When the population distribution is normal, the sampling distribution of $\bar{y}$ is exactly normal for any sample size.

## Central Limit Theorem (sum form)
Let $X_1, X_2, \ldots, X_n$ be independent and identically distributed random variables
with finite mean $\mu = \mathbb{E}[X_i]$ and finite variance $\sigma^2 = \mathrm{Var}(X_i)$.

## Define the random sum
## $
S_n = \sum_{i=1}^{n} X_i .
$

Then, as $n \to \infty$,
## $\frac{S_n - n\mu}{\sqrt{n\sigma^2}}
\;\xrightarrow{d}\;
\mathcal{N}(0,1).$

Equivalently, for large $n$,
## $S_n \approx \mathcal{N}(n\mu,\; n\sigma^2).$


### Note: The Central Limit Theorem applies to the sum of random variables (or their realizations in repeated experiments), not to the sum of their means.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Reproducibility
np.random.seed(42)

# Exponential distribution parameter
lam = 1.0                 # rate parameter
mu = 1 / lam              # mean
sigma2 = 1 / lam**2       # variance

# CLT parameters
n_size = 1000                    # number of RVs in each sum
n_trials = 100000          # number of repeated experiments

# Draw samples: shape = (n_trials, n)
samples = np.random.exponential(scale=1/lam, size=(n_trials, n_size))

# Sum of realizations
S_n = samples.sum(axis=1)

# CLT-normalized sum
Z = (S_n - n_size * mu) / np.sqrt(n_size * sigma2)

# Plot
plt.figure(figsize=(7,5))
plt.hist(Z, bins=70, density=True, alpha=0.7, label='CLT-normalized sums')

# Standard normal PDF
x = np.linspace(-4, 4, 500)
plt.plot(x, norm.pdf(x), 'r', lw=2, label='Standard Normal PDF')

plt.xlabel(r'$(S_n - n\mu)/\sqrt{n\sigma^2}$')
plt.ylabel('Probability density')
plt.title('Central Limit Theorem: Exponential Distribution')
plt.legend()
plt.show()
```

# Covariance

For two jointly distributed real-valued random variables $X$ and $Y$ with finite second moments, the covariance is defined as the expected value (or mean) of the product of their deviations from their individual expected values:

# $Cov(X,Y) = E[(X −E[X])(Y −E[Y])] = E[XY] − E[X]E[Y]$.

When Y=X i.e

# $Cov(X,X) = Var(X) = \sigma_X^2$


<!-- For a pair of random variables, $X$ and $Y$, their covariance is defined as

# $Cov[X, Y] = E[(X − E[X])(Y − E[Y])] = E[XY] − E[X]E[Y]$ -->

A covariance is defined for each pair of random variables in the distribution. Say, there are $p$ number of random variates then there will be $p(p-2)/2$ covariances.

The covariance provides a measure of the correlation between the different random variables in the distribution. A nonzero value will thus indicate a relation of some sort between the two variables, while the sign of the value will indicate the sense of the relation; i.e., positive or negative.

If the variables are independent, the covariance is zero. That is for two independent variables, $E[X,Y] = E[X]\, E[Y]$. Substituting this in the above equation will give $Cov[X,Y] = 0$.

However, the converse is not necessarily true. That is if the covariance of X and Y are zero, it need not mean that they are independent random variates. For example, lets consider $Y = X^2$. Then the distribution of X is symmetric around $0$ and so, $E[x] =0$. Similarly, $E[X,Y] = E[X^3]$ is also symmetric around $0$. Therefore, the expection of $X^3$ is zero. So, in the above equation, we will get the $Cov[X,Y] =0$, but $X$ and $Y$ are dependent variables here.


Covariance is also expressed in terms of correlation coefficient,

# $ \rho_{X,Y} = \frac{Cov[X,Y]}{\sigma_x \, \sigma_y}$

# This is also referred to as Pearson Correlation Coefficient (PCC) or Pearson's product moment correlation coefficient or simply correlation coefficient.

It is a measure of $\bf{linear \, correlation}$ between two sets of data. The correlation coefficient can take on values between $-1$ and $+1$, the sign of which indicates the sense of the correlation. And obviously, the correlation of a variable with itself is $\rho_{X,X} = 1$.



(Credits:Denis Boigelot)

For a set of two variables **X** and **Y**, the covariance matrix ($\Sigma$) is:

$\Sigma =
\begin{bmatrix}
\text{Var}(X) & \text{Cov}(X, Y) \\
\text{Cov}(Y, X) & \text{Var}(Y)
\end{bmatrix}
$

Where:
- $\text{Var}(X)$ is the variance of **X**,
- $\text{Var}(Y)$ is the variance of **Y**,
- $\text{Cov}(X, Y)$ (or  $\text{Cov}(Y, X)$) is the covariance between **X** and **Y**.


The diagonal elements are the variances of the random variates and the off-diagonal elements presents the covariance between the various pairs of the random variates.

The covariance matrix is always symmetric semi-definite matrix (the eigen values of the matrix are non-negative). If the variables are mutually independent, then the matrix reduces to a diagonal matrix.

The covariance matrix depends on the scale of the values. That is if one multiplies a factor to each of the random variates, then the values of the covariance matrix will change.

**Why is it important?**

The covariance matrix is especially useful in statistics, machine learning, and data analysis, as it helps to understand the relationships between variables. For example, it is a key component in Principal Component Analysis (PCA) for dimensionality reduction, where the matrix tells you the directions of greatest variance in the data.

```python
#Defining a covariance Matrix
def covariance(X):
    n = len(X)
    cov = np.array([0]*(n*n)).reshape(n,n) #Initialising the matrix
    cov = np.array([[(np.multiply(X[i],X[j])).mean() - X[i].mean() *X[j].mean() for j in range(n)] for i in range(n)],dtype=float)

    return np.array(cov)

#Example
np.set_printoptions(precision=2)
X1 = np.linspace(1,10,10)
X2 = np.linspace(5,15,10)
X3 = np.linspace(-1,5,10)
X =[X1,X2,X3]
print('The covariance matrix is \n',covariance(X))
print(np.var(X1),np.var(X2),np.var(X3))
```

```python
def Pearson_correlation(X):
    n=len(X)

    P_corr_coeff = ((np.multiply(X[0],X[1])).mean() - X[0].mean() *X[1].mean())/(np.std(X[0])*np.std(X[1]))

    return np.array(P_corr_coeff)

X_ = np.array([[1, 2, 3, 4, 5], [10, 9, 2.5, 6, 4]])

print('Pearson linear correlation coefficient is',Pearson_correlation(X_))

#Numpy function
print(np.corrcoef(X_))

#Scipy function
from scipy import stats

res = stats.pearsonr(X_[0], X_[1])
print(res)
```

# Spearman’s Correlation Coefficient

Spearman's rank correlation coefficient assess the montonic relation between two datasets using the ranking of the two datasets

# $r_s = \rho_{R_X,R_Y} =\frac{Cov(R(X)R(Y))}{\sigma_{R(X)} \sigma_{R(Y)}} $




## Interpretation of the Spearman Correlation Coefficient:

### $r_s=1:$ A perfect positive monotonic relationship. As one variable increases, the other increases in a perfectly predictable manner (but not necessarily linearly).

### $r_s=−1$: A perfect negative monotonic relationship. As one variable increases, the other decreases in a perfectly predictable manner.

### $r_s=0$: No monotonic relationship. This indicates that the ranks of the two variables are not related in any way.

### $0<r_s<1$: A positive monotonic relationship, meaning that as one variable increases, the other tends to increase as well, but not necessarily at a constant rate.

### $−1<r_s<0$: A negative monotonic relationship, meaning that as one variable increases, the other tends to decrease, but again, not necessarily at a constant rate.

```python
from scipy import stats
import numpy as np

def create_ranks(X):
    array = np.array(X)
    T = array.argsort() #using argsort get indices of values arranged in ascending order
    ranks = np.empty_like(T)
    ranks[T] = np.arange(len(array))
    return(ranks)


def Spearman_correlation(X):
    RX = create_ranks(X[0])
    RY = create_ranks(X[1])
    print(RX,RY)

    S_corr_coeff = ((np.multiply(RX,RY)).mean() - RX.mean() *RY.mean())/(np.std(RX)*np.std(RY))

    return S_corr_coeff


X = [4,5,10,23,56]
Y = [23,4,89,34,56]
X_=np.array([X,Y])

print('The Spearman correlation coefficient of the given pair of dataset is ',Spearman_correlation(X_))


plt.figure()
plt.scatter(X,Y)
plt.tick_params(labelsize=15)
plt.tick_params(axis='x', labelsize=18)
plt.tick_params(axis='y', labelsize=18)
plt.show()

```
