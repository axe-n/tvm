# Measurements and Errors
Ref: Statistical Methods in physical sciences

All experiments involve making measurements using some type of instrument. The uncertainties involved in these measurements are mainly of two types:

 # 1. Systematic Errors

Systematic errors concern the possible biases that may be present in an observation. Systematic errors arise whenever there is a comparison between two or more measurements. For example, in a balance or a voltmeter, the measurement is made with reference to the 'zero point'. Even if greatest care is taken, one can never be fully certain that the instrument is exactly at zero point before the measurement. This is due to various physical factors, like the thickness of the scale lines, lighting available during the calibration, eyesight of the calibrator etc. As a result, one can only say that the instrument has been zeroed to within some range of values, $0 \pm \delta$. This uncertainty in the zero point will then bring about a bias in the subsequent measurements made by the instrument.

More scientific example, in detecting the presence of a new particle, the signal must be compared to the background events that could simulate such a particle. The art of experimentation is to minimise these errors.
The major part of the effort in experiments, therefore, is to identify the possible sources of systematic error, design the experiment so as to minimize or eliminate as many of these as possible and monitor those that remain!

Differentiating between systematic and random errors, when compared between different measurements, the systematic error will remain constant whereas the random errors will keep changing in a completely independent fashion. Random errors can be reduced by increased number of repeated measurements, however, this is not the case for systematic errors.

Another point to be noted is that in different stages of data analysis, a random error in one stage can become a systematic error in other stages. For example, in the calibrating process of fixing the zero of the balance or the voltmeter, the error, $\delta$ we get is a random error. However, for later measurements, this becomes the systematic error.   





# 2. Random Errors

The source of these uncertainties can be instrumental imprecisions or the inherent random nature of the process being measured or both.

-- $\bf{Instrumental \, Errors}:$ These errors arise from the inability of the observer and the measuring instrument to control all the variables that can affect the measurement. For most instruments, we can argue, in fact, that probability distribution function of errors, f(e) is a Gaussian with mean 0 and a variance $\sigma$.

Indeed, if $e$ represents the sum of many random factors, then, the central limit theorem ensures that f(e) will tend towards a normal distribution, regardless of the distributions of the individual factors. In such a case the standard deviation obtained for the distribution, $\sigma$ corresponds to the precision of the instrument.

-- $\bf{Statistical \, Errors}:$ A second source of random error arises when the process being observed or measured is itself a random or stochastic process. The measurement uncertainty incurred can be identified as the sampling error. Because they result from the randomness of the process itself, these are also called staristical errors.


Let us try to model this simple situation. Suppose $x$ is the true value that we are seeking to measure and suppose we make $n$ repeated measurements, $x$, with the same instrument. What is observed, then, are values $y_i$

# $y_i = x + s + e_i \, (i = 1, 2, 3,....,n)$

where $s$ is the systematic error, $e_i$ is the random error distributed according to some probability density function, $f(e)$. Essentially, the true value, $x$ is unobservable since it is always accompanied by the error, $e$. And indeed all that can be measured directly is the distribution $f(e)$, offset by the value $x$ and the systematic error. The measurements are thereby just random sample from this distribution.


If $e$ represents the sum of many random factors, then, the central limit theorem ensures that $f(e)$ will tend towards a normal distribution, regardless of the distributions of the individual factors. The expected value of $y_i$ is then $E(X) = x$, which is the true value being sought.


# Example: Modeling a simple realistic experiment
A realistic experiment, say
### the counting of the number of disintegrations in a period, $\Delta t$, from a radioactive source with a detector.
There will be two sources of events: one from the radioactive source itself and the other from background sources such as cosmic rays, radioactive elements in the surroundings, or radioactive contaminants in the source itself. As a first simplification let us assume that there is only one $\gamma$-ray line, so that the energy does not vary. Let $n_s$ be the number of events emitted by the radioactive source in time $\Delta t$ and $n_b$ the number from background sources in the same period. The number of these events, in both cases, is governed
by the Poisson distribution

# $p_{src} = \frac{(\lambda_s \Delta t)^{n_s} e^{-\lambda_s \Delta t}}{n_s!}$

# $p_{bkg} = \frac{(\lambda_b \Delta t)^{n_b} e^{-\lambda_b \Delta t}}{n_b!}$

where $\lambda_s$ and $\lambda_b$ are the mean decay rates for the source and background respectively. This, however, is not what is observed since the detector has an efficiency, $\epsilon$. This efficiency includes the geometric solid angle subtended by the detector relative to the source as well as its intrinsic detection efficiency, the latter being the probability that the detector will actually register an event when a gamma ray enters into its active volume. As a second simplification, let us assume $\epsilon$ to be a constant.

The probability that the detector will register $n_r$ counts when $n_s + n_b$ are emitted is a “Bernoulli trials” situation. So, the respective binomial distribution is given by

# $p_{det}(n_r|n_s,n_b) = \frac{(n_s + n_b)!}{n_r! (n_s+n_b - n_r)!} \epsilon^{n_r} (1-\epsilon)^{n_s+n_b - n_r}$


Now, we can describe the experiment which is the joint probability of

# $p(n_r, n_s,n_b) = p_{det}(n_r|n_s,n_b) \times p_{src} \times p_{bkg}$

This in turn allows us to estimate the marginal probability

# $p_{obs}(n_r) = \sum_{n_s} \sum_{n_b} p(n_r, n_s,n_b)$

This mathematical probability describes our experiment telling us what is the probability to observe, $n_r$ photons in the detector.


Let us have a look at some of the expected values. The expected number of counts is then

# $ E(n_r) = \sum_{n_r} n_r p_{obs}(n_r) = \sum_{n_r} \sum_{n_s} \sum_{n_b} n_r p_{det}(n_r|n_s,n_b) p_{src}(n_s)p_{bkg}(n_b)$

# $= \epsilon \, (\lambda_s + \lambda_b) \, \Delta t$

The variance,
# $ Var(n_r) = E(n_r^2) - E^2(n_r) = \epsilon \, (\lambda_s + \lambda_b) \, \Delta t$

Since the mean and variance of the $n_r$ distribution is the same, the distribution of $n_r$ is a Poisson.

By making a series of measurements we can thus estimate $E(n_r)$ and its error. To obtain the mean of the signal rate, one must know the efficiency of the detector $\epsilon$ and the background rate $\lambda_b$.

We need to conduct independent additional experiments (or Monte Carlo simulations) to constrain these two parameters which are eventually required to estimate the mean and variance of the average number of signal detected.

# Error Propagation or Propagation of uncertainties

Using the independent measurements of the various parameters we later on estimate the dependent variable. The random error estimates on the independent parameters now needs to be propagated in order to estimate the error in the dependent quantity.

Lets define function, $y$ such that

# $y = f(A,B)$

The error in $y$ is given as
# $\sigma_y = \sqrt{\frac{\sum (y-\bar{y})^2}{n}}$

The Taylor expansion of the $y$ at $(\bar{a},\bar{b})$ gives:

# $y- \bar{y} \sim (A - \bar{a}) \frac{\partial f}{\partial A} + (B - \bar{b}) \frac{\partial f}{\partial B}$

Substituting in the above equation, we get

# $ \sigma_y^2 = \frac{1}{n} \sum \left[ (A - \bar{a}) \frac{\partial f}{\partial A} +  (B - \bar{b}) \frac{\partial f}{\partial B} \right]^2 $

# $ \sigma_y^2 = \frac{1}{n} \sum \left[ (A - \bar{a})^2 \left(\frac{\partial f}{\partial A}\right)^2 +  (B - \bar{b})^2 \left(\frac{\partial f}{\partial B}\right)^2 + 2 (A - \bar{a}) (B - \bar{b}) \frac{\partial f}{\partial A}\frac{\partial f}{\partial B}  \right]$

# $ \sigma_y^2 = \sigma_A^2 \left(\frac{\partial f}{\partial A}\right)^2 + \sigma_B^2 \left(\frac{\partial f}{\partial B}\right)^2 + 2 cov(A,B)\frac{\partial f}{\partial A}\frac{\partial f}{\partial B}$

```python
import uncertainties as unc
import uncertainties.unumpy as unumpy
import numpy as np
import matplotlib.pyplot as plt

# Defines x and y
x=T90
y=fluence

# Defines the error arrays, values follow a normal distribution
errx=T90_err
erry=fluence_err

# Defines special arrays holding the values *and* errors
X=unumpy.uarray((x, errx ))
Y=unumpy.uarray((y, erry ))

Z = (Y/X)
#Print the value of x/y
z = unumpy.nominal_values(np.array(Z))
# Print the propagated error errz
errz=unumpy.std_devs(Z)
#print(errz)

plt.figure()
plt.errorbar(x,y,xerr=errx,yerr=erry,fmt='r^',alpha=0.08)
plt.tick_params(labelsize=15)
plt.tick_params(axis='x', labelsize=18)
plt.tick_params(axis='y', labelsize=18)
plt.yscale('log')
plt.xscale('log')interpolate.
plt.xlabel('T90 (s)',fontsize=20)
plt.ylabel('Fluence (erg/cm2)',fontsize=20)
plt.show()



plt.figure()
#plt.errorbar(x,y,xerr=errx,yerr=erry,fmt='ko')
plt.errorbar(x,z,xerr=errx,yerr=errz,fmt='ko',alpha=0.08)
plt.tick_params(labelsize=15)
plt.tick_params(axis='x', labelsize=18)
plt.tick_params(axis='y', labelsize=18)
plt.yscale('log')
plt.xscale('log')
plt.xlabel('T90 (s)',fontsize=20)
plt.ylabel('Flux (erg/cm2/s)',fontsize=20)
plt.show()
```

# Propagating errors by implementing the expression

# $ \sigma_y^2 = \sigma_A^2 \left(\frac{\partial f}{\partial A}\right)^2 + \sigma_B^2 \left(\frac{\partial f}{\partial B}\right)^2 + 2 cov(A,B)\frac{\partial f}{\partial A}\frac{\partial f}{\partial B}$

```python
def covariance(X):
    n = len(X)
    cov = np.array([0]*(n*n)).reshape(n,n) #Initialising the matrix
    cov = np.array([[(np.multiply(X[i],X[j])).mean() - X[i].mean() *X[j].mean() for j in range(n)] for i in range(n)],dtype=float)

    return np.array(cov)


import sympy as sym
from sympy import *
from scipy.stats import norm
x_mean = 4e-3
y_mean = 2e1
x_sigma = 1e-2
y_sigma = 5

x_ = norm.rvs(loc=x_mean, scale=x_sigma,size=1000)
y_ = norm.rvs(loc=y_mean, scale=y_sigma,size=1000)

x, y = symbols('x y')
z = x/y
print("Expression : {} ".format(z))

# Use sympy.Derivative() method

print("Derivative of expression with respect to x : {}".format(sym.diff(z,x)))
f_x = sym.diff(z, x).evalf(subs={x: x_mean, y: y_mean})
print("Derivative of z with respect to x at x=x_mean, y= y_mean : {}".format(f_x))

print("Derivative of expression with respect to y : {}".format(sym.diff(z,y)))
f_y = sym.diff(z, y).evalf(subs={x: x_mean, y: y_mean})
print("Derivative of z with respect to y at x=x_mean, y= y_mean : {}".format(f_y))
cov_xy = covariance([x_,y_])[0,1]
print('Covariance between the two quantities is',cov_xy)

sigma_z = (x_sigma**2 * f_x**2 + y_sigma**2 * f_y**2 + 2* cov_xy * f_x * f_y)**0.5

print('The standard deviation of z is',sigma_z)

```
