# (b) Probability Histogram

## <span style="color: Maroon;"> Note on Probability Density in Histograms <span>

When plotting a histogram with `density=True`, the heights of the bins represent a probability density function. The scaling ensures that the total area under the histogram integrates to 1.

The density for each bin is calculated as:
`density[i] = counts[i] / (sum(counts) * bin_width[i])`

where:np.diff(bins)[i]
- `counts[i]:` Raw count for the i-th bin
- `sum(counts):` Total number of counts across all bins
- `bin_width[i]:` Width of the i-th bin, calculated as `np.diff(bins)[i]`

### <span style="color: Maroon;"> Normalisation Condition:<span>

The area under the histogram is normalized such that:
`np.sum(density * np.diff(bins)) == 1`
This ensures that the histogram properly represents a probability density function.








```python
# Creating Probability density histogram
fig2, ax = plt.subplots(figsize =(5, 4))
bin_width = 5
ax.hist(mass, bins = round((np.max(mass) - np.min(mass))/bin_width),density=True)
# Show plot
ax.set_xlabel('Mass of stars in solar masses',fontsize=15)
ax.set_ylabel('Probability density',fontsize=15)
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)
plt.show()
```

```python
# getting data of the histogram
bin_width=5
count, bins_edges = np.histogram(mass, bins=round((np.max(mass) - np.min(mass))/bin_width))

# Probability / relative frequency distribution
pb = count / sum(count)

# the function numpy np.cumsum allows to calculate the cumulative probability/ relative frequency
cpb = np.cumsum(pb)

# plotting PDF and CDF
%matplotlib inline
fig, ax = plt.subplots(figsize =(5,3))
plt.tight_layout()
plt.step(bins_edges[1:], pb, color=
         "red", label="relative frequency", where='mid') # bins_edges[1:] --> all bin edges except the first one (upper bounds) and bins_edges[:-1] --> all bin edges except the last one (lower bounds)
plt.step(bins_edges[1:], cpb, label="cumulative relative frequency", where='mid')
plt.yscale('log')
plt.xlabel('Mass of stars in solar masses',fontsize=15)
plt.ylabel('Probability',fontsize=15)
plt.tick_params(axis='x', labelsize=12)
plt.tick_params(axis='y', labelsize=12)
plt.legend(loc='center right')
ax.set_ylim([0.001,1.5])
plt.show()
```

# Mode
The mode of a set of observations/measurements of a variable is defined as that value of the variable that has the highest frequency.

$\bf{Characteristics}$
1. It is the most frequent or probable measurement in the data set.
2. There can be more than one mode for a data set.
3. It is not influenced by extreme measurements.
4. Modes of subsets cannot be combined to determine the mode of the complete data set.
5. For grouped data its value can change depending on the categories used.
6. It is applicable for both qualitative and quantitative data.



```python
from scipy import stats
mode_fluence = stats.mode(fluence)
print(mode_fluence)  #return values [array of mode values, Array of counts for each mode]
mode_T90 = stats.mode(T90)
print(mode_T90)
print(len(T90))
```

Note in the above examples, calculating the mode using the ungrouped data is not a good idea as the values are fractional and highly sensitive to minor changes in the data.
In such cases, the best way to calculate the mode of the data is using the grouped data (first doing a histogram and making a grouped data format)
First identify the interval with the highest frequency --> modal class. Then use the below formula to calculate the mode

$Mode = L + \left ( \frac{f_m - f_1}{(f_m - f_1) + (f_m - f_2)}\right) h$


where,

L = lower boundary of the modal class

$f_m$ = frequency of the modal class

$f_1$ = frequency of the class preceding the modal class

$f_2$ = frequency of the class following the modal class

h = width of the class interval

```python
import numpy as np

# Example ungrouped (fractional) data
data = np.array([
    12.84, 13.12, 13.35, 13.58, 13.74,
    13.91, 14.05, 14.18, 14.32, 14.47,
    14.59, 14.73, 14.86, 14.98, 15.11,
    15.23, 15.36, 15.48, 15.61, 15.74,
    15.88, 16.01, 16.15, 16.28, 16.42,
    16.57, 16.71, 16.86, 17.01, 17.17,
    17.33, 17.50, 17.68, 17.87, 18.07,
    18.28, 18.50, 18.73, 18.97, 19.22,
    19.48, 19.75, 20.03, 20.32, 20.62,
    20.93, 21.25, 21.58, 21.92, 22.27
])


# 1. Generate histogram (grouped data)
freq, bin_edges = np.histogram(data, bins=10)

# Build grouped data table
grouped_data = pd.DataFrame({
    "Class Interval": [
        f"{bin_edges[i]:.2f} – {bin_edges[i+1]:.2f}"
        for i in range(len(freq))
    ],
    "Frequency": freq
})

print(grouped_data)

# 2. Plot a histogram
fig2, ax = plt.subplots(figsize =(5, 4))
ax.hist(data, bins = 10,density=False)
# Show plot
ax.set_xlabel('Random Data',fontsize=15)
ax.set_ylabel('Counts per bin',fontsize=15)
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)
plt.show()

# 3. Identify modal class index
modal_idx = np.argmax(freq)

# 4. Extract parameters for mode formula
fm = freq[modal_idx]
f1 = freq[modal_idx - 1] if modal_idx > 0 else 0
f2 = freq[modal_idx + 1] if modal_idx < len(freq) - 1 else 0

L = bin_edges[modal_idx]                  # lower boundary of modal class
h = bin_edges[1] - bin_edges[0]           # class width

# 5. Grouped-data mode formula
mode = L + ((fm - f1) / ((fm - f1) + (fm - f2))) * h

print("Grouped-data mode:", mode)
print(stats.mode(data))
```

### For ungrouped data with decimal (continuous) values or discrete (whole-number) data, if most values occur only once or very few times, the ungrouped mode becomes unstable or uninformative. In such cases, grouping the data into suitable class intervals and estimating the mode from the grouped frequency distribution provides a more meaningful description of the data’s most typical values.

### Mode is more appropriate for data types like Categorical (nominal) data like blood group, gender etc, Discrete numerical data with clear repetition like number of children per family, defects per item etc.

# Median
 The median of a set of observations/measurements of a variable is defined as the middle value when the values of the variable are arranged from lowest to highest (ascending/descending order).

Depending on whether the dataset has an odd or even number of elements.

1. **For an odd number of data points**:  
   The median is the middle value in the sorted dataset. There is exactly one value that is in the center.

   - **Example**:  
     For the dataset `[1, 3, 5]`, the median is **3** because it is the middle value.

2. **For an even number of data points**:  
   The median is calculated by averaging the two middle values in the sorted dataset.

   - **Example**:  
     For the dataset `[1, 3, 5, 7]`, the two middle values are 3 and 5. The median is the average of these two values:  
     $\text{Median} = \frac{3 + 5}{2}= 4$
     
     

 $\bf{Characteristics}$

1. It is the central value; 50% of the measurements lie above it and 50% fall below it.
2. There is only one median for a data set.
3. It is not influenced by extreme measurements.
4. Medians of subsets cannot be combined to determine the median of the complete data set.
5. For grouped data, its value is rather stable even when the data are organized into different categories.
6. It is applicable to quantitative data only.

$\bf{How \, to \, find \, Median \, for \, a \, grouped \, data \, (output\, of \, a \, histogram)?}$

It is generally difficult because we do not know the actual values of the variable. Thereby, it is difficult to find in which class interval the median lies.
If we assume that the measurements are spread evenly throughout the interval, we get the following result. Let
- L =lower class limit of the interval that contains the median
- N = total frequency
- c $f_b$ = the sum of frequencies (cumulative frequency) for all classes before the median class
- $f_m$ = frequency of the class interval containing the median
- w = interval width

Then, for grouped data, the median is given by

$ median = L + \frac{w}{f_m} (.5N - cf_b)$

$\bf Steps\,  to \, Calculate \, Median\,  for \, Grouped \, Data$
1. Organise the data: Ensure you have class intervals and their frequencies.
2. Calculate the cumulative frequencies.
3. Find the median class: The class where the cumulative frequency first exceed N/2.
4. Apply the formula

# Mean
 The arithmetic mean, or mean of a set of observations is defined to be the sum of the values of the variable divided by the total number of values.

 $\bf{Characteristics}$
1. It is the arithmetic average of the measurements in a data set.
2. There is only one mean for a data set.
3. Its value is influenced by extreme measurements; trimming can help to reduce the degree of influence.
4. Means of subsets can be combined to determine the mean of the complete data set.
5. It is applicable to quantitative data only.

$\bf{Trimmed \, Mean}$

The presence one or more extreme values can distort the estimate of mean. In such situations, a variation of mean can be used which is called as trimmed mean. This is estimated by dropping the highest and lowest extreme values and averages the rest. For example, a $5\%$ trimmed mean drops the highest $5\%$ and the lowest $5\%$ of the measurements and averages the rest.

$\bf{Weighted Mean}$

In this case, all the data points do not equally contribute to the estimation of the mean. There is a weight associated to each of the data. The weights take into account the varying degree of importance that is associated with each data of the dataset.

Sample mean of a grouped data,

$ \frac{ \sum_{i} f_iy_i}{N}$

where,

- $f_i$ is the frequency
- $y_i$ is the variable value
- N is the total frequency (total no: of data)

# Measures of Variability in the Data

In the below Figure


even though all the different distributions tend to have same mean, each of them have different spread or variability about their mean.

- $\bf{Range}$

The range of a set of measurements is defined to be the difference between the largest and the smallest measurements of the set.

In the case of the grouped data, the range is considered to be from the lower limit of the first class interval and the upper limit of the last class interval. The range is sensitive to outliers because it depends on the most extreme values and it does not tell us about the variability of the distribution of the values about its mean.


- $\bf{Percentiles}$

The $p^{th}$ percentile of a set of $n$ measurements arranged in order of magnitude is that value that has at most p% of the measurements below it and at most (100 - p)% above it.




Some of the percentiles that are of interest are $25\%$, $50\%$, $75\%$ which are also referred to as lower quartile, middle quartile, upper quartile.



- $\bf{Deviation}$

The deviations of the measurements are computed by using the formula $y$ - $\bar{y}$ where $\bar{y}$ is the arithmetic mean.



An average of all the deviations does not make sense as that will tend to zero. Therefore, an interpretable quantity is the average of the sum of the squared deviations of the measurements from their mean which is called as ${\bf variance}$.

- $\bf{Variance}$

The variance of a set of $n$ measurements y1, y2, . . . , yn with mean $\bar{y}$ is the sum of the squared deviations divided by $n - 1$:

# $ \frac{ \sum_{i} (y_i - \bar{y})^2}{n-1}$

For grouped data,
# $ \frac{ \sum_{i} f_i (y_i - \bar{y})^2}{n-1}$

Lets say the sample variance to be $s^2$ and the population variance is represented by $\sigma^2$. The use of (n - 1) as the denominator of $s^2$ is not arbitrary. This definition of the sample variance makes it an unbiased estimator of the population variance $\sigma^2$. This means roughly that if we were to draw a very large number of samples, each of size n, from the population of interest and if we computed $s^2$ for each sample, the average sample variance would equal the population variance $\sigma^2$. Had we divided by n in the definition of the sample variance $s^2$, the average sample variance computed from a large number of samples would be less than the population variance; hence, $s^2$ would tend to underestimate $\sigma^2$.

- ${\bf Standard \, Deviation}$

The standard deviation of a set of measurements is defined to be the positive square root of the variance. One reason for defining the standard deviation is that it yields a measure of variability having the same units of measurement as the original data, whereas the units for variance are the square of the measurement units.

Among the different measures of variability, the standard deviation is particularly appealing for two reasons: (1) we can compare the variabilities of two or more sets of data using the standard deviation, and (2) we can also use the results of the rule that follows to interpret the standard deviation of a single set of measurements. This rule applies to data sets with roughly a histogram that has a single peak, is symmetrical, and tapers off gradually in the tails. Because so many data sets can be classified as mound-shaped, the rule has wide applicability. For this reason, it is called the Empirical Rule.




$\bf{Empirical \, rule}$

Give a set of n measurements possessing a mound-shaped/ symmetric histogram, then the interval
- $y \pm s$ contains approximately 68% of the measurements
- $y \pm 2s$ contains approximately 95% of the measurements
- $y \pm 3s$ contains approximately 99.7% of the measurements

### Boxplot

A **boxplot** (also known as a **box-and-whisker plot**) is a graphical representation of the distribution of a dataset, highlighting its central tendency, spread, and potential outliers. It is particularly useful for comparing distributions between different groups or identifying the presence of outliers.

#### Components of a Boxplot:
1. **Box**:
   - Represents the **interquartile range (IQR)**, which contains the middle 50% of the data. The box extends from the **first quartile (Q1)** to the **third quartile (Q3)**, with the width of the box reflecting the spread of the central data.
   
2. **Median Line**:
   - A line inside the box representing the **median (Q2)**, or the middle value of the dataset.

3. **Whiskers**:
   - Lines extending from the box that represent the range of data within a specified range, typically 1.5 times the IQR from Q1 and Q3. The whiskers show the extent of the data excluding outliers.
   
4. **Outliers**:
   - Data points that fall outside the whiskers, often plotted as individual points. These are typically defined as values more than 1.5 times the IQR away from Q1 and Q3.

#### Key Features:
- **Central tendency**: The median line shows the central value of the dataset.
- **Spread**: The box shows how data is spread around the median, with the length of the box indicating the IQR.
- **Outliers**: Points outside the whiskers are considered outliers, indicating values that are significantly higher or lower than most of the data.

#### Usage:
Boxplots are widely used for:
- **Identifying skewness**: If the median line is not in the center of the box, the data may be skewed. Thus, a skewed histogram produces an asymmetric box plot, where the box and whiskers stretch more in the direction of the tail.
- **Comparing distributions**: Useful for comparing multiple datasets side by side.
- **Detecting outliers**: Outliers are easily identifiable as individual points outside the whiskers.

### Examples of Right and Left Skewed Box Plots

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

right_skewed = np.random.exponential(scale=1.0, size=1000)

plt.figure()
plt.boxplot(right_skewed, vert=False)
plt.title("Right-Skewed (Positively Skewed) Box Plot")
plt.xlabel("Value")
plt.show()
```

```python
left_skewed = -np.random.exponential(scale=1.0, size=1000)

plt.figure()
plt.boxplot(left_skewed, vert=True)
plt.title("Left-Skewed (Negatively Skewed) Box Plot")
plt.xlabel("Value")
plt.show()
```
